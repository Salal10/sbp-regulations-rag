```python
from pathlib import Path
from pypdf import PdfReader

RAW_DIR = Path("data/raw")

for path in sorted(RAW_DIR.glob("*.pdf")):
    reader = PdfReader(path)
    page_texts = [page.extract_text() or "" for page in reader.pages]
    chars = [len(text.strip()) for text in page_texts]
    empty_pages = sum(1 for c in chars if c < 50)
    print(f"{path.name:42} pages={len(chars):3}  chars={sum(chars):7,}  empty_pages={empty_pages}")
```

```python
sample = PdfReader(RAW_DIR / "complaint_handling_guidelines_2016.pdf").pages[0].extract_text()
print(sample[:1500])
```
